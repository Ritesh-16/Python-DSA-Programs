```python
#Queue using 2 stack
# 1 stack for pushing and other stack for emptying 
# so it behaves as a queue
```

```python
class QueueUsingStacks:
    def __init__(self):
        self.stack1 = []
        self.stack2 = []

    def enqueue(self,value):
        self.stack1.append(value)
    
    def dequeue(self):
        if not self.stack2:
            if not self.stack1:
                return "Empty Queue"
            while self.stack1:
                self.stack2.append(self.stack1.pop())
        if not self.stack2:
            return "Empty Queue"
        return self.stack2.pop()
        
```

```python
q = QueueUsingStacks()
q.enqueue(1)
q.enqueue(2)
q.enqueue(3)
print(q.dequeue())  # Output: 1
print(q.dequeue())  # Output: 2
print(q.dequeue())
```
