#Queue using Linked List

```python
#Since it is using Linked List, it requires Node
class Node:
    def __init__(self,value):
        self.value = value
        self.next = None
```

```python
class queue:
    def __init__(self):
        self.rear = None
        self.front = None
    
    def enqueue(self,value):
        new_node = Node(value)
        if self.rear == None:
            self.front = new_node
            self.rear = self.front
        else:
            self.rear.next = new_node
            self.rear = new_node

    def dequeue(self):
        if self.front == None:
            return "Empty Queue"
        else:
            self.front = self.front.next

    def traverse(self):
        if self.front == None:
            return "Empty Queue"
        curr = self.front
        while curr!= None:
            print(curr.value,end=' ')
            curr = curr.next
    
    def is_empty(self):
        return self.front == None
    
    def size(self):
        curr = self.front
        size = 0
        while curr!= None:
            size += 1
            curr = curr.next
        return size
    def front_item(self):
        if self.front == None:
            return "Empty"
        return self.front.data
    
    def rear_item(self):
        if self.front == None:
            return "Empty"
        return self.rear.data

q = queue()
```

```python
q.enqueue(4)
q.enqueue(16)
q.enqueue(9)
q.enqueue(2)
```

```python
q.dequeue()
```
